```python
import random

import numpy as np

from mdpforge import Benchmark

DISCOUNT = 0.9
SIZE = "large"
PRECISION = 1e-3
REPEATS = 2
SEED = 0

TIMEOUT = 20  # Secondes par essai ; None désactive l'isolation en processus.

np.random.seed(SEED)
random.seed(SEED)

bench = Benchmark()
bench.reset()
bench.add_solver('mdpforge_mpi')
bench.add_all_mdps(size=SIZE)
results = bench.run(
    discount=DISCOUNT,
    precision=PRECISION,
    repeats=REPEATS,
    seed=SEED,
    timeout=TIMEOUT,
    verbose=True,
)
```
