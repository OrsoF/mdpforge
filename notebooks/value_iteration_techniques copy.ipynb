```python
import random

import numpy as np

from mdpforge import Benchmark

DISCOUNT = 0.9
SIZE = "medium"
PRECISION = 1e-3
REPEATS = 3
SEED = 0

TIMEOUT = 5  # Secondes par essai ; None désactive l'isolation en processus.

np.random.seed(SEED)
random.seed(SEED)
```

```python
bench = Benchmark()
bench.reset()
bench.add_solver('mdpforge_mpi')
bench.add_all_mdps(size=SIZE)
```

```python
results = bench.run(
    discount=DISCOUNT,
    precision=PRECISION,
    repeats=REPEATS,
    seed=SEED,
    timeout=TIMEOUT,
    verbose=True,
)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(50, 8))
results.plot_heat(ax=ax)
```

```python
results.plot_performance_profile(max_factor=20)
```

```python
results.summary()
```

```python
results.export_csv()
```
