# Aggregated methods

## Total reward: small models

```python
from mdpforge import Benchmark

SIZE = "small"

bench = Benchmark(criterion="total")

bench.add_all_mdps(size=SIZE)
# bench.add_mdp(['rooms'])

bench.add_solver("mdpforge_vi")
bench.add_solver("mdpforge_qvi")
bench.add_solver("mdpforge_mpi")
bench.add_solver("mdpforge_aggvi")
bench.add_solver("mdpforge_aggqvi")
bench.add_solver("mdpforge_aggmpi")
bench.add_solver("mdpforge_tvi")

result = bench.run(timeout=10)

result.plot_heat(reference="mdpforge_aggqvi")
result.plot_performance_profile()
```

## Discounted reward: aggregation threshold

```python
from mdpforge import Benchmark

DISCOUNT = 0.99
SIZE = "small"
REPEATS = 3
TIMEOUT = 5

bench = Benchmark()
bench.add_all_mdps(size=SIZE)

for switch_to_vi_ratio in [0.5, 0.8, 0.9, 0.95, 0.99, 0.999]:
    bench.add_solver("mdpforge_aggqvi", switch_to_vi_ratio=switch_to_vi_ratio)


result_small = bench.run(
    discount=DISCOUNT,
    repeats=REPEATS,
    timeout=TIMEOUT,
    verbose=True,
)

result_small.plot_performance_profile(max_factor=2)
```
