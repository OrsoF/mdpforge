```python
from mdpforge import Benchmark

DISCOUNT = 0.99
SIZE = "tiny"

bench = Benchmark()
bench.add_mdp("rooms", size=SIZE)
bench.add_mdp("mountain", size=SIZE)
bench.add_mdp("taxi", size=SIZE)
bench.add_mdp("sysadmin", size=SIZE)

bench.add_mdp("tandem", size=SIZE)
bench.add_mdp("dam", size=SIZE)
bench.add_mdp("inventory", size=SIZE)
bench.add_mdp("impatience", size=SIZE)

bench.add_mdp("garnet", size=SIZE)
bench.add_mdp("block", size=SIZE)

bench.add_solver("mdpforge_vi")
bench.add_solver("mdpforge_qvi")
bench.add_solver("mdpforge_mpi")
bench.add_solver("mdpforge_aggvi")
bench.add_solver("mdpforge_aggqvi")
bench.add_solver("mdpforge_aggmpi")
bench.add_solver("mdpforge_chen")
bench.add_solver("mdpforge_bertsekas")
bench.add_solver("mdpforge_abate")

result = bench.run(DISCOUNT)

result.plot_heat()
result.plot_performance_profile()
```

```python
from mdpforge import Benchmark

SIZE = "large"

bench = Benchmark(criterion='total')

bench.add_mdp(["rooms", 'mountain', 'taxi', 'sysadmin', 'frozen'], size=SIZE)

bench.add_solver("mdpforge_vi")
bench.add_solver("mdpforge_qvi")
bench.add_solver("mdpforge_mpi")
bench.add_solver("mdpforge_aggvi")
bench.add_solver("mdpforge_aggqvi")
bench.add_solver("mdpforge_aggmpi")
bench.add_solver("mdpforge_tvi")

result = bench.run()

result.plot_heat()
result.plot_performance_profile()
```
