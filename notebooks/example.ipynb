# Minimal benchmark example

```python
from mdpforge import Benchmark

DISCOUNT = 0.8

bench = Benchmark()
bench.add_default_mdps()
bench.add_solver("mdpforge_mpi")
bench.add_solver("mdpforge_vi")
result = bench.run(DISCOUNT)
```

```python
_ = result.plot_elo()
```

```python
_ = result.plot_pairwise_wins()
```

```python
_ = result.plot_status()
```

```python
_ = result.plot_runtime_distribution()
```
